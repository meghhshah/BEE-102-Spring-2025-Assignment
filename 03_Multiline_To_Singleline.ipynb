### Multi-line sequence fasta to single-line fasta file.

The multi-line input file is in the folder. This code will create a new file for single-line output.

```python
input_file = "multiline_input.fasta"
output_file = "singleline_output.fasta"
with open(input_file, 'r') as fin, open(output_file, 'w') as fout:
    sequence = ""
    for line in fin:
        line = line.strip()
        if line.startswith('>'):  # this line is a header
            if(sequence):
                fout.write(sequence + "\n")
                sequence = ""
            fout.write(line + "\n")
        else:
            sequence += line
    
    # write the last sequence
    fout.write(sequence + "\n")

print("The output file is ready.")
```
